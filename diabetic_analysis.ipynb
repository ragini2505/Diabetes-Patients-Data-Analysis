```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt  # Corrected import
import seaborn as sns

# Load data
df = pd.read_csv("diabetese.tsv", sep="\t")
fig, ax = plt.subplots(figsize=(15, 15))

# Create a boxplot
sns.boxplot(data=df, ax=ax)

# Save the boxplot as an image
plt.savefig("boxplot.png")

# Calculate the correlation matrix
correlation_matrix = df.corr()
print(correlation_matrix)  # Optionally print the correlation matrix
plt.figure(figsize=(15, 15))  
ax = sns.heatmap(df.corr(), annot=True)

# Save the heatmap as an image file
plt.savefig("correlation_heatmap.png")  # Specify the filename and format (e.g., .png)

plt.show()  # Display the heatmap



```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt  
import seaborn as sns

# Load data
df = pd.read_csv("diabetese.tsv", sep="\t")


sns.pairplot(df)
plt.title("Scatter plots for all pairs of variables in the DataFrame")
plt.show() 
plotImg=sns.pairplot(df)
plotImg.savefig("dataframe.png")




```

```python
#According to body mass index
import matplotlib.pyplot as plt 
df['BMI'].hist(bins=12)
plt.title("BMI Distribution Among Diabetic Patients")
plt.xlabel("BMI Range")
plt.ylabel("Count")
plt.show()
```

```python
#I want to understand what age of the patients are mostly diabetic : Histogram
import matplotlib.pyplot as plt 
df['AGE'].hist(bins=12)
plt.title("Age Distribution Among Diabetic Patients")
plt.xlabel("Age Range")
plt.ylabel("Count")
plt.show()
```

```python
#Boxplotting for BMI
import matplotlib.pyplot as plt 
plt.figure(figsize=(6,2))
plt.boxplot(df['BP'],vert=False,showmeans=True)
plt.grid(color='gray',linestyle='dotted')
plt.xlabel("Value")
plt.ylabel("BP")
plt.show()
```

```python
#Box plotting for BMI
import matplotlib.pyplot as plt 
plt.figure(figsize=(6,2))
plt.boxplot(df['BMI'],vert=False,showmeans=True)
plt.grid(color='gray',linestyle='dotted')
plt.xlabel("Value")
plt.ylabel("BMI")
plt.show()
```

```python
#Boxplotting
import matplotlib.pyplot as plt  
plt.figure(figsize=(2,6))

plt.boxplot(df['AGE'],showmeans=True)
plt.grid(color='gray',linestyle='dotted')
plt.xlabel("Age")
plt.ylabel("years")
plt.title("Boxplotting")
plt.show()
```

```python
#Standard deviation
std_age=df['AGE'].std()
print(f"Standard deviation is : {std_age}")
```

```python
#variance
var_age=df['AGE'].var()
print(f"The variance is : {var_age} ")
```

```python
#Mean age of diabetic patients
mean_age=df['AGE'].mean()
print(f"The mean age of diabetese patients is : {mean_age}")
```
